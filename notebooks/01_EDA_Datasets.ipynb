```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
municipios = pd.read_csv(
    "../datasets/municipios/municipios.csv",
    encoding="latin-1",
    sep=";"
)

# MUNICIPIOS

# Número de filas
print("Número de filas: ", municipios.shape)

# Tipos de columnas
print("Tipos de columnas: ", municipios.info())

# Primeros registros
print("Primeros registros: ", municipios.head())

# Estadísticas
print("Estadísticas: ", municipios.describe())

# Valores nulos
print("Valores nulos: ", municipios.isnull().sum())

# Valores duplicados
print("Valores duplicados: ", municipios.duplicated().sum())

# Nombre de las columnas
print("Nombre de las columnas: ", municipios.columns)
```
