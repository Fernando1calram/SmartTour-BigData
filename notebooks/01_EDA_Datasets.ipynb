```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
municipios = pd.read_csv(
    "../datasets/municipios/municipios.csv",
    encoding="latin-1",
    sep=";"
)

# MUNICIPIOS

# Número de filas
print("=====================================================")
print("Número de filas:")
print(municipios.shape)
print("=====================================================")

# Tipos de columnas
print("=====================================================")
print("Tipos de columnas:")
print(municipios.info())
print("=====================================================")

# Primeros registros
print("=====================================================")
print("Primeros registros:")
print(municipios.head())
print("=====================================================")

# Estadísticas
print("=====================================================")
print("Estadísticas:")
print(municipios.describe())
print("=====================================================")

# Valores nulos
print("=====================================================")
print("Valores nulos: ")
print(municipios.isnull().sum())
print("=====================================================")

# Valores duplicados
print("=====================================================")
print("Valores duplicados: ")
print(municipios.duplicated().sum())
print("=====================================================")

# Nombre de las columnas
print("=====================================================")
print("Nombre de las columnas: ")
print(municipios.columns)
print("=====================================================")
```
