```python
import os
import pandas as pd

def cargar_grupos(ruta_carpeta):
    alumnos = set()
    for archivo in os.listdir(ruta_carpeta):
        if(archivo.endswith(".csv")):
            ruta = os.path.join(ruta_carpeta, archivo)
            df = pd.read_csv(ruta)

            alumnos.update(df["nombre"].dropna().astype(str))
    return alumnos
```

```python
grupo_a = cargar_grupos("../../datasets/grupo_a")
grupo_b = cargar_grupos("../../datasets/grupo_b")
```

```python
# -----------------------------
# 3. ANÁLISIS BÁSICO
# -----------------------------
print("Jugadores únicos Club A:", len(grupo_a))
print("Jugadores únicos Club B:", len(grupo_b))

print("\nListado Club A (ordenado):")
print(sorted(grupo_a))

print("\nListado Club B (ordenado):")
print(sorted(grupo_b))
```

```python
# -----------------------------
# 4. OPERACIONES DE CONJUNTOS
# -----------------------------

# Unión
union = grupo_a | grupo_b
print("\nUNIÓN:", len(union))

# Intersección
interseccion = grupo_a & grupo_b
print("\nINTERSECCIÓN:", len(interseccion))
print(sorted(interseccion))

# Diferencia A - B
solo_a = grupo_a - grupo_b
print("\nSOLO CLUB A:", len(solo_a))

# Diferencia B - A
solo_b = grupo_b - grupo_a
print("\nSOLO CLUB B:", len(solo_b))

# Diferencia simétrica
diferencia_simetrica = grupo_a ^ grupo_b
print("\nDIFERENCIA SIMÉTRICA:", len(diferencia_simetrica))

# Complemento de la intersección respecto al universo
complemento_interseccion = union - interseccion
print("\nCOMPLEMENTO INTERSECCIÓN:", len(complemento_interseccion))

"""union = grupo1 | grupo2
intersección = grupo1 & grupo2
diferencia = grupo1 - grupo2
diferencia_simetrica = grupo1 ^ grupo2"""
```
