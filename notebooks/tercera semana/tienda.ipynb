```python
import pandas as pd

# Ventas (CSV)
ventas = pd.read_csv("../../datasets/clientes/ventas.csv")

# Productos (Excel)
productos = pd.read_excel("../../datasets/clientes/productos.xlsx")

# Opiniones (JSON)
opiniones = pd.read_json("../../datasets/clientes/opiniones.json")

# Visualización de datos
"""print(ventas)
print(productos)
print(opiniones)"""

# valores nulos
"""print("Ventas is null")
print(ventas.isnull().sum())
print("Productos is null")
print(productos.isnull().sum())
print("Opiniones is null")
print(opiniones.isnull().sum())"""

# duplicados
print("DUPLICATED")
print("Ventas duplicated")
print(ventas.duplicated().sum())
print("Productos duplicated")
print(productos.duplicated().sum())
print("Opiniones duplicated")
print(opiniones.duplicated().sum())
```

```python
# Unión de ventas con productos
ventas_productos = pd.merge(ventas, productos, on="producto_id")
# How como el join
# ventas_productos = pd.merge(ventas, productos, on="producto_id", how="left")
print(ventas_productos)
```

```python
# Unión de opiniones con clientes
datos_completos = pd.merge(ventas_productos, opiniones, on="cliente_id")
#datos_completos = pd.merge(ventas_productos, opiniones, on="cliente_id", how="left")
print(datos_completos)
```

```python
# Total gastado por cliente
datos_completos["total"] = datos_completos["cantidad"] * datos_completos["precio"]

gasto_cliente = datos_completos.groupby("cliente_id")["total"].sum()
print(gasto_cliente)
```

```python
# Valoración media por cliente
valoracion_media = datos_completos.groupby("cliente_id")["valoracion"].mean()
print(valoracion_media)
```

```python
# Producto más vendido
producto_popular = datos_completos["nombre"].value_counts()
print(producto_popular)
```
